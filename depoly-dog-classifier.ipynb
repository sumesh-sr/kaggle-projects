```python
import ipywidgets as widgets
from fastai.learner import load_learner
import warnings
from fastai.vision.all import *
```

```python
warnings.filterwarnings("ignore", category=UserWarning, module="fastai")

model = load_learner('/kaggle/input/models/sumeshsr/dog-classifier/other/default/1/export.pkl')
```

```python
btn_run = widgets.Button(description='Classify')
btn_upload = widgets.FileUpload()
out = widgets.Output()
lbl_pred = widgets.Label()
```

```python
def on_click_classify(change):
    img=PILImage.create(btn_upload.value[0].content.tobytes())
    out.clear_output()
    with out:display(img.to_thumb(128,128))
    pred,pred_idx,probs=model.predict(img)
    lbl_pred.value = f'Prediction: {pred}; Probability: {probs[pred_idx]:.04f}'
btn_run.on_click(on_click_classify)
```

```python
widgets.VBox([widgets.Label('Select dog!'), 
      btn_upload, btn_run, out, lbl_pred])
```
